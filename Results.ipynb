```python
from pathlib import Path

import matplotlib.patches as mpatches
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.lines import Line2D
from matplotlib.ticker import FixedLocator, FuncFormatter

from src.Results.PlotsBestModels import PlotsBestModels


# ============================================================
# CONFIGURAÇÃO VISUAL
# ============================================================

plt.rcParams.update(
    {
        "font.family": "serif",
        "font.serif": [
            "Times New Roman",
            "Times",
            "DejaVu Serif",
        ],
        "font.size": 8.0,
        "axes.titlesize": 7.8,
        "xtick.labelsize": 7.0,
        "ytick.labelsize": 7.0,
        "legend.fontsize": 6.8,
        "figure.dpi": 200,
        "savefig.dpi": 400,
        "pdf.fonttype": 42,
        "ps.fonttype": 42,
    }
)


pastasModelos = [
    "HoeffdingTree",
    "HoeffdingAdaptiveTree",
    "AdaptiveRandomForest",
    "LeveragingBagging",
    "AdaptiveIsolationForest",
    "HalfSpaceTrees",
    "Autoencoder",
]


cenarios = [
    "Consistency_1000",
    "Generalization_1000",
    "Adaptation_200",
    "Recurrence_200",
]


configuracaoCenarios = {
    "Consistency_1000": {
        "nome": "Consistency",
        "bloco": "1k/block",
    },
    "Generalization_1000": {
        "nome": "Generalization",
        "bloco": "1k/block",
    },
    "Adaptation_200": {
        "nome": "Adaptation",
        "bloco": "200/block",
    },
    "Recurrence_200": {
        "nome": "Recurrence",
        "bloco": "100/block",
    },
}


coresErros = {
    "FP": "#0072B2",
    "FN": "#D62728",
}


coresAtaques = {
    "DNS": "#E69F00",
    "SYN": "#009E73",
    "LDAP": "#CC79A7",
}


aquecimento = 2000


# ============================================================
# LOCALIZAÇÃO DOS RESULTADOS
# ============================================================

todosCsv = {}

for pasta in pastasModelos:
    pastaResultados = Path("output", pasta)

    if not pastaResultados.exists():
        continue

    for caminho in pastaResultados.rglob("*.csv"):
        partes = [
            parte.lower()
            for parte in caminho.parts
        ]

        if "prequential" not in partes:
            todosCsv[str(caminho)] = None


if not todosCsv:
    raise FileNotFoundError(
        "Nenhum CSV foi encontrado nas pastas output/<modelo>."
    )


# ============================================================
# METADADOS DOS ATAQUES
# ============================================================

plots = PlotsBestModels(
    metadata_path="data/attack_regions_metadata.json"
)


metadata = plots.load_attack_metadata()


if not metadata:
    metadata = plots.build_attack_metadata(
        data_root="data/15k",
        label_col="Label",
        output_path="data/attack_regions_metadata.json",
    )


def agruparBlocos(
    regioes,
    intervaloMaximo=500,
):
    blocos = []

    regioesOrdenadas = sorted(
        regioes,
        key=lambda item: item["start"],
    )

    for regiao in regioesOrdenadas:
        atual = {
            "start": int(regiao["start"]),
            "end": int(regiao["end"]),
            "attack_name": str(
                regiao["attack_name"]
            ).upper(),
        }

        mesmoBloco = (
            blocos
            and blocos[-1]["attack_name"]
            == atual["attack_name"]
            and atual["start"]
            - blocos[-1]["end"]
            - 1
            <= intervaloMaximo
        )

        if mesmoBloco:
            blocos[-1]["end"] = max(
                blocos[-1]["end"],
                atual["end"],
            )
        else:
            blocos.append(atual)

    return blocos


# ============================================================
# SELEÇÃO DOS MELHORES MODELOS
# ============================================================

candidatos = plots.build_candidates_dataframe(
    todosCsv,
    window_size=100,
)


selecionados = plots.select_best_models(
    candidatos,
    top_n=1,
)


selecionados = selecionados[
    selecionados["English_Dataset"].isin(cenarios)
].copy()


quantidades = (
    selecionados
    .groupby("English_Dataset")["Task_Type"]
    .nunique()
    .reindex(cenarios)
    .fillna(0)
)


if quantidades.lt(2).any():
    faltantes = quantidades[
        quantidades.lt(2)
    ].index.tolist()

    raise RuntimeError(
        "Não foi encontrado um classificador e um "
        "detector de anomalias para: "
        + ", ".join(faltantes)
    )


# ============================================================
# CARREGAMENTO DAS SÉRIES PREQUENCIAIS
# ============================================================

execucoes = {}


for cenario in cenarios:
    for tipo in (
        "classifier",
        "anomaly",
    ):
        resultado = selecionados[
            (
                selecionados["English_Dataset"]
                == cenario
            )
            & (
                selecionados["Task_Type"]
                == tipo
            )
        ]

        if resultado.empty:
            raise RuntimeError(
                f"Resultado ausente para {cenario} ({tipo})."
            )

        linha = resultado.iloc[0]

        serie = (
            plots
            ._read_prequential_series(linha)
            .sort_values("Instance")
            .reset_index(drop=True)
        )

        if serie.empty:
            raise RuntimeError(
                f"Série vazia para {cenario} ({tipo})."
            )

        execucoes[(cenario, tipo)] = {
            "linha": linha,
            "serie": serie,
        }


blocosPorCenario = {
    cenario: agruparBlocos(
        metadata[cenario]["attack_regions"]
    )
    for cenario in cenarios
}


# ============================================================
# LIMITES DOS EIXOS
# ============================================================

maiorInstancia = max(
    np.nanmax(
        execucao["serie"][
            "Instance"
        ].to_numpy(dtype=float)
    )
    for execucao in execucoes.values()
)


limiteX = (
    -maiorInstancia * 0.025,
    maiorInstancia * 1.015,
)


def calcularLimiteYGlobal():
    maioresValores = []

    for execucao in execucoes.values():
        serie = execucao["serie"]

        x = serie[
            "Instance"
        ].to_numpy(dtype=float)

        mascaraAvaliacao = (
            x >= aquecimento
        )

        for metrica, desvio in (
            ("FP_avg", "FP_std"),
            ("FN_avg", "FN_std"),
        ):
            media = serie[
                metrica
            ].to_numpy(dtype=float)

            if desvio in serie.columns:
                sd = (
                    serie[desvio]
                    .fillna(0)
                    .to_numpy(dtype=float)
                )
            else:
                sd = np.zeros_like(media)

            limiteSuperior = media + sd

            mascaraValida = (
                mascaraAvaliacao
                & np.isfinite(limiteSuperior)
            )

            if mascaraValida.any():
                maioresValores.append(
                    np.nanmax(
                        limiteSuperior[mascaraValida]
                    )
                )

    if not maioresValores:
        return 10

    maiorValor = max(maioresValores)

    if maiorValor <= 10:
        passo = 2
    elif maiorValor <= 25:
        passo = 5
    elif maiorValor <= 50:
        passo = 10
    else:
        passo = 20

    return min(
        100,
        max(
            passo,
            np.ceil(maiorValor / passo) * passo,
        ),
    )


limiteSuperiorY = calcularLimiteYGlobal()


margemInferiorY = max(
    2.5,
    limiteSuperiorY * 0.10,
)


limiteYGrafico = (
    -margemInferiorY,
    limiteSuperiorY,
)


marcasY = np.linspace(
    0,
    limiteSuperiorY,
    5,
)


marcasX = list(
    np.arange(
        0,
        maiorInstancia + 1,
        4000,
    )
)


def formatarInstancia(
    valor,
    posicao,
):
    if np.isclose(valor, 0):
        return "0"

    if abs(valor) >= 1000:
        valorMilhares = valor / 1000

        if float(valorMilhares).is_integer():
            return f"{int(valorMilhares)}k"

        return f"{valorMilhares:.1f}k"

    return f"{valor:.0f}"


def formatarErro(
    valor,
    posicao,
):
    if np.isclose(valor, round(valor)):
        return str(int(round(valor)))

    return f"{valor:.1f}"


def compactarNomeModelo(linha):
    nome = plots._series_label(linha)

    substituicoes = {
        "Hoeffding Adaptive Tree": "HAT",
        "HoeffdingAdaptiveTree": "HAT",
        "Hoeffding Tree": "HT",
        "HoeffdingTree": "HT",
        "Adaptive Random Forest": "ARF",
        "AdaptiveRandomForest": "ARF",
        "Leveraging Bagging": "LB",
        "LeveragingBagging": "LB",
        "Adaptive Isolation Forest": "AIF",
        "AdaptiveIsolationForest": "AIF",
        "Half-Space Trees": "HST",
        "Half Space Trees": "HST",
        "HalfSpaceTrees": "HST",
        "Autoencoder": "AE",
        "Full Features": "All Features",
        "Full features": "All Features",
        "All features": "All Features",
        "Best Features": "Selected Features",
        "Best features": "Selected Features",
        "Selected features": "Selected Features",
        " - ": " . ",
        " · ": " . ",
        " | ": " . ",
    }

    for antigo, novo in substituicoes.items():
        nome = nome.replace(
            antigo,
            novo,
        )

    return nome


# ============================================================
# DESENHO DOS PAINÉIS
# ============================================================

def desenharPainel(
    eixo,
    cenario,
    tipo,
    letraPainel,
):
    execucao = execucoes[
        (cenario, tipo)
    ]

    linha = execucao["linha"]
    serie = execucao["serie"]

    configuracao = configuracaoCenarios[
        cenario
    ]

    x = serie[
        "Instance"
    ].to_numpy(dtype=float)

    mascaraAvaliacao = (
        x >= aquecimento
    )

    # Warm-up indicado somente pelo fundo cinza.
    eixo.axvspan(
        0,
        aquecimento,
        facecolor="#BDBDBD",
        alpha=0.35,
        linewidth=0,
        zorder=1,
    )

    # Regiões dos ataques.
    for indice, bloco in enumerate(
        blocosPorCenario[cenario]
    ):
        inicio = max(
            bloco["start"],
            aquecimento,
        )

        fim = min(
            bloco["end"],
            maiorInstancia,
        )

        if inicio >= fim:
            continue

        nomeAtaque = bloco[
            "attack_name"
        ]

        corAtaque = coresAtaques.get(
            nomeAtaque,
            "#777777",
        )

        eixo.axvspan(
            inicio,
            fim,
            facecolor=corAtaque,
            alpha=0.22,
            linewidth=0,
            zorder=1,
        )

        nivelTexto = (
            0.96
            if indice % 2 == 0
            else 0.80
        )

        eixo.text(
            (inicio + fim) / 2,
            nivelTexto,
            nomeAtaque,
            transform=eixo.get_xaxis_transform(),
            ha="center",
            va="top",
            fontsize=6.0,
            fontweight="bold",
            color=corAtaque,
            bbox={
                "facecolor": "white",
                "edgecolor": corAtaque,
                "linewidth": 0.30,
                "alpha": 0.88,
                "boxstyle": "round,pad=0.06",
            },
            clip_on=True,
            zorder=10,
        )

    # Linhas de FP e FN.
    for sigla, metrica, desvio in (
        ("FP", "FP_avg", "FP_std"),
        ("FN", "FN_avg", "FN_std"),
    ):
        media = serie[
            metrica
        ].to_numpy(dtype=float)

        if desvio in serie.columns:
            sd = (
                serie[desvio]
                .fillna(0)
                .to_numpy(dtype=float)
            )
        else:
            sd = np.zeros_like(media)

        mascaraValida = (
            mascaraAvaliacao
            & np.isfinite(x)
            & np.isfinite(media)
            & np.isfinite(sd)
        )

        xValido = x[
            mascaraValida
        ]

        mediaValida = media[
            mascaraValida
        ]

        sdValido = sd[
            mascaraValida
        ]

        eixo.fill_between(
            xValido,
            np.maximum(
                mediaValida - sdValido,
                0,
            ),
            np.minimum(
                mediaValida + sdValido,
                limiteSuperiorY,
            ),
            color=coresErros[sigla],
            alpha=0.09,
            linewidth=0,
            zorder=3,
        )

        eixo.plot(
            xValido,
            mediaValida,
            color=coresErros[sigla],
            linewidth=0.7,
            solid_capstyle="round",
            zorder=6,
        )

    nomeModelo = compactarNomeModelo(
        linha
    )

    tipoTitulo = (
        "Classifier"
        if tipo == "classifier"
        else "Anomaly Detector"
    )

    titulo = (
        f"{tipoTitulo}\n"
        f"{configuracao['nome']} . "
        f"{configuracao['bloco']}\n"
        f"({letraPainel}) {nomeModelo}"
    )

    eixo.set_title(
        titulo,
        pad=3.5,
        fontsize=7.8,
        fontweight="semibold",
        linespacing=1.08,
    )

    eixo.set_xlim(
        *limiteX
    )

    eixo.set_ylim(
        *limiteYGrafico
    )

    eixo.xaxis.set_major_locator(
        FixedLocator(marcasX)
    )

    eixo.xaxis.set_major_formatter(
        FuncFormatter(
            formatarInstancia
        )
    )

    eixo.yaxis.set_major_locator(
        FixedLocator(marcasY)
    )

    eixo.yaxis.set_major_formatter(
        FuncFormatter(
            formatarErro
        )
    )

    eixo.tick_params(
        axis="both",
        which="major",
        pad=1.8,
        length=2.8,
        width=0.70,
    )

    eixo.grid(
        True,
        linestyle=":",
        linewidth=0.55,
        alpha=0.38,
        zorder=0,
    )

    for borda in eixo.spines.values():
        borda.set_linewidth(
            0.70
        )


# ============================================================
# CRIAÇÃO DA FIGURA 2 × 4
# ============================================================

figura, eixos = plt.subplots(
    2,
    4,
    figsize=(7.16, 4.00),
    sharex=True,
    sharey=True,
    facecolor="white",
)


letrasClassificadores = [
    "a",
    "b",
    "c",
    "d",
]


letrasDetectores = [
    "e",
    "f",
    "g",
    "h",
]


for coluna, cenario in enumerate(
    cenarios
):
    desenharPainel(
        eixos[0, coluna],
        cenario,
        "classifier",
        letrasClassificadores[coluna],
    )

    desenharPainel(
        eixos[1, coluna],
        cenario,
        "anomaly",
        letrasDetectores[coluna],
    )


# Valores do eixo Y somente na primeira coluna.
for eixo in eixos[:, 1:].flat:
    eixo.tick_params(
        axis="y",
        labelleft=False,
    )


# Valores do eixo X nas duas linhas.
for eixo in eixos.flat:
    eixo.tick_params(
        axis="x",
        labelbottom=True,
    )


figura.supxlabel(
    "Stream instance",
    x=0.535,
    y=0.095,
    fontsize=8.5,
    fontweight="semibold",
)


figura.supylabel(
    "Errors per 100 instances",
    x=0.012,
    y=0.51,
    fontsize=8.5,
    fontweight="semibold",
)


# ============================================================
# LEGENDA GLOBAL
# ============================================================

elementosLegenda = [
    Line2D(
        [0],
        [0],
        color=coresErros["FP"],
        linewidth=1.0,
        label="FP",
    ),
    Line2D(
        [0],
        [0],
        color=coresErros["FN"],
        linewidth=1.0,
        label="FN",
    ),
    mpatches.Patch(
        facecolor="#BDBDBD",
        alpha=0.45,
        label="Warm-up (0–2k)",
    ),
    mpatches.Patch(
        facecolor=coresAtaques["DNS"],
        alpha=0.40,
        label="DNS",
    ),
    mpatches.Patch(
        facecolor=coresAtaques["SYN"],
        alpha=0.40,
        label="SYN",
    ),
    mpatches.Patch(
        facecolor=coresAtaques["LDAP"],
        alpha=0.40,
        label="LDAP",
    ),
]


figura.legend(
    handles=elementosLegenda,
    loc="lower center",
    bbox_to_anchor=(0.5, 0.012),
    ncol=6,
    frameon=False,
    columnspacing=1.25,
    handlelength=1.8,
    handletextpad=0.42,
)


# ============================================================
# ESPAÇAMENTO FINAL
# ============================================================

figura.subplots_adjust(
    left=0.072,
    right=0.995,
    top=0.91,
    bottom=0.19,
    hspace=0.72,
    wspace=0.10,
)


# ============================================================
# SALVAMENTO
# ============================================================

pastaSaida = Path(
    "output/Results/plots_paper"
)


pastaSaida.mkdir(
    parents=True,
    exist_ok=True,
)


arquivoPdf = pastaSaida / (
    "BestModels_FP_FN_AllScenarios_2x4.pdf"
)


arquivoPng = pastaSaida / (
    "BestModels_FP_FN_AllScenarios_2x4.png"
)


figura.savefig(
    arquivoPdf,
    format="pdf",
    bbox_inches="tight",
    pad_inches=0.025,
    facecolor="white",
)


figura.savefig(
    arquivoPng,
    format="png",
    dpi=400,
    bbox_inches="tight",
    pad_inches=0.025,
    facecolor="white",
)


plt.show()
plt.close(figura)


print(f"Saved: {arquivoPdf}")
print(f"Saved: {arquivoPng}")
```
